```python
### Libaries
import sklearn
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import fetch_california_housing, fetch_openml
from sklearn.tree import DecisionTreeClassifier, DecisionTreeRegressor
from sklearn.ensemble import BaggingClassifier, ExtraTreesClassifier, RandomForestClassifier, AdaBoostClassifier, HistGradientBoostingClassifier, GradientBoostingRegressor, VotingClassifier, StackingClassifier
from sklearn.metrics import accuracy_score, mean_squared_error
```

```python
### Random Seed
RANDOM_SEED = 0
```

```python
### Data
fashionMNIST = fetch_openml('Fashion-MNIST')
californiaHousing = fetch_california_housing()
```

```python
### Part 1: Single CART trees

depths = [2,4,6,8,10,12,15,20,None]

for depth in depths:
    # create trees
    entropyTree = DecisionTreeClassifier(criterion='entropy', random_state=RANDOM_SEED, max_depth=depth)
    giniTree = DecisionTreeClassifier(criterion='gini', random_state=RANDOM_SEED, max_depth=depth)

    # fit and predict for both
    entropyTree.fit(fashionMNIST.data, fashionMNIST.target)
    entropyTreePredictions = entropyTree.predict(fashionMNIST.data)

    giniTree.fit(fashionMNIST.data, fashionMNIST.target)
    giniTreePredictions = giniTree.predict(fashionMNIST.data)

    # show depth and accuracy
    print("Depth: {0:.4f}".format(depth=depth))

    entropyAccuracy = accuracy_score(fashionMNIST.target, entropyTreePredictions)
    giniAccuracy = accuracy_score(fashionMNIST.target, giniTreePredictions)

    print("Entropy Tree Accuracy: {0:.4f}".format(entropyAccuracy))
    print("Gini Tree Accuracy: {0:.4f}".format(giniAccuracy))
```
